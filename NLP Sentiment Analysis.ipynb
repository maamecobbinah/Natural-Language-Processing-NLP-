**Problem Statement**

“Our media company wants to analyze audience reviews of their shows to automatically classify them as positive or negative. 
Build an NLP model that predicts sentiment based on the text of the review.”

Step-by-Step Approach

1. **Data collection** – Collect text reviews and their sentiment labels (positive/negative).
2. **Data preprocessing** – Clean text: lowercasing, removing punctuation, stopwords, tokenization.
3. **Feature extraction** – Convert text to numeric representation (e.g., TF-IDF or embeddings).
4. **Model selection** – Use a classifier (e.g., Logistic Regression, Random Forest, or a small neural network).
5. **Training and evaluation** – Train model, evaluate using metrics like accuracy, F1-score.
6. **Prediction / deployment** – Use the model to classify new reviews.

```python
# Step 0: Imports
import pandas as pd
import re
import string
```

```python
data = pd.DataFrame({
    'review': [
        "I loved the show! Amazing story and acting.",
        "Terrible acting, I hated it.",
        "It was okay, some good parts but overall meh.",
        "Absolutely fantastic, can't wait for more episodes!",
        "Boring and predictable, not worth it.",
        "Great visuals and engaging plot!",
        "The storyline was confusing and dull.",
        "Loved the characters, very relatable.",
        "I fell asleep, extremely boring.",
        "Excellent pacing and fantastic dialogues.",
        "Not impressed, the show lacked depth.",
        "Amazing soundtrack and performances.",
        "Disappointing, I expected much better.",
        "Wonderful plot twists, kept me hooked.",
        "Poorly written script, very predictable.",
        "Brilliant acting, highly recommend!",
        "Waste of time, very dull and slow.",
        "Incredible storytelling and cinematography.",
        "Mediocre performance, not enjoyable.",
        "One of the best shows I've watched this year!",
        "The series was terrible, wouldn't recommend.",
        "Engaging and entertaining, loved it.",
        "The plot was weak and uninteresting.",
        "Fantastic direction and amazing cast.",
        "Awful, I couldn't finish the episode."
    ],
    'sentiment': [
        "positive", "negative", "negative", "positive", "negative",
        "positive", "negative", "positive", "negative", "positive",
        "negative", "positive", "negative", "positive", "negative",
        "positive", "negative", "positive", "negative", "positive",
        "negative", "positive", "negative", "positive", "negative"
    ]
})

print(data.head())
```

```python
print(data.describe())
print(data.info())
```

```python
# Step 2: Data Preprocessing
def preprocess_text(text):
    text = text.lower()  # lowercase
    text = re.sub(f"[{string.punctuation}]", "", text)  # remove punctuation
    text = re.sub(r"\d+", "", text)  # remove numbers
    return text

data['cleaned_review'] = data['review'].apply(preprocess_text)
```

```python
# Step 3: Train-test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    data['cleaned_review'], 
    data['sentiment'], 
    test_size=0.2, 
    random_state=42, 
    stratify=data['sentiment']  # ensures both classes appear
)
```

```python
# Step 4: Feature extraction (TF-IDF)
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
# Step 5: Model training
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train_tfidf, y_train)
```

```python
# Step 6: Predictions
y_pred = model.predict(X_test_tfidf)
```

```python
# Step 7: Evaluation
from sklearn.metrics import accuracy_score, classification_report
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
# Step 8: Predicting new reviews
new_reviews = ["The show was fantastic and very exciting!", "I did not like it, very dull."]
new_reviews_cleaned = [preprocess_text(r) for r in new_reviews]
new_reviews_tfidf = vectorizer.transform(new_reviews_cleaned)
predictions = model.predict(new_reviews_tfidf)
print("\nPredictions:", list(zip(new_reviews, predictions)))
```
